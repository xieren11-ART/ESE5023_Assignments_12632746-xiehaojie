```python
#第三题代码
def Pascal_triangle(k):
    num = 1
    for j in range(k):
        print(num,end='')
        num = num * (k - 1 - j) // (j + 1)

Pascal_triangle(100)
```

```python
# 第四题代码
import random
x = random.randint(1, 100)
print(x)
def Least_moves(x):
    time = 0

    while x > 1:
        if x % 2 == 0:
            x = x / 2
        else:
            x = x - 1

        time = time + 1

    print(time)

Least_moves(x)
```

```python
#第五题代码 5.1
def Find_expression(x):
    count = 0

    for code in range(6561):
        n = code
        total = 0
        number = 1
        sign = 1
        expression = "1"

        for digit in range(2, 10):
            choice = n % 3
            n = n // 3

            if choice == 0:
                number = number * 10 + digit
                expression = expression + str(digit)

            else:
                total = total + sign * number
                number = digit

                if choice == 1:
                    sign = 1
                    expression = expression + "+" + str(digit)

                else:
                    sign = -1
                    expression = expression + "-" + str(digit)

        total = total + sign * number

        if total == x:
            print(expression, "=", x)
            count = count + 1

    return count


Find_expression(10)
```

```python
# 第五题 5.2
def Find_expression(x, show=True):
    count = 0

    for code in range(6561):
        n = code
        total = 0
        number = 1
        sign = 1
        expression = "1"

        for digit in range(2, 10):
            choice = n % 3
            n = n // 3

            if choice == 0:
                number = number * 10 + digit
                expression = expression + str(digit)

            else:
                total = total + sign * number
                number = digit

                if choice == 1:
                    sign = 1
                    expression = expression + "+" + str(digit)
                else:
                    sign = -1
                    expression = expression + "-" + str(digit)

        total = total + sign * number

        if total == x:
            if show:
                print(expression, "=", x)
            count = count + 1

    return count
import matplotlib.pyplot as plt

Total_solutions = []

for i in range(1, 101):
    count = Find_expression(i, False)
    Total_solutions.append(count)

plt.figure(figsize=(12, 5))
plt.plot(range(1, 101), Total_solutions)

plt.xlabel("Target number")
plt.ylabel("Number of solutions")
plt.title("Number of solutions for each target")
plt.grid(True)
plt.show()

maximum = max(Total_solutions)
minimum = min(Total_solutions)

print("最大表达式数量：", maximum)
print("对应的整数：", [i + 1 for i in range(100) if Total_solutions[i] == maximum])

print("最小表达式数量：", minimum)
print("对应的整数：", [i + 1 for i in range(100) if Total_solutions[i] == minimum])
```
